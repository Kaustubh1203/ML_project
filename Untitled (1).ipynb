#### Step 2: Feature Engineering`

##### 2.1 Age Group

```python
# Create Age Group

bins = [18, 30, 45, 60, 100]

labels = ["Young", "Adult", "Middle Age", "Senior"]

df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

##### 2.2 Income Category

```python
# Create Income Category
bins = [0, 300000, 600000, 1000000, float("inf")]

labels = [
    "Low Income",
    "Medium Income",
    "High Income",
    "Very High Income"
]

df["income_category"] = pd.cut(
    df["annual_income"],
    bins=bins,
    labels=labels
)
```

#### Step 2.5: Payment Delay Risk Feature

```python
# Create Payment Delay Risk Score
df["payment_delay_risk"] = (
    df["late_payments_last_year"]
    +
    (2 * df["missed_payments_last_year"])
    +
    (df["avg_days_to_pay_after_due"] / 10)
    +
    df["grace_period_used_count"]
)
```

```python
# Verify New Feature
df[[
    "late_payments_last_year",
    "missed_payments_last_year",
    "avg_days_to_pay_after_due",
    "grace_period_used_count",
    "payment_delay_risk"
]].head(10)
```

```python
# Create Risk Categories
def payment_risk_category(x):

    if x <= 3:
        return "Low Risk"

    elif x <= 10:
        return "Medium Risk"

    else:
        return "High Risk"


df["payment_risk_category"] = (
    df["payment_delay_risk"]
    .apply(payment_risk_category)
)
```

##### 5.2 INFORMATION VALUE (IV)


##### 5.2.1 Function to Calculate IV


```python
def calculate_iv(data, feature, target, bins=10):

    temp = data[[feature, target]].copy()

    # Convert numerical feature into bins
    temp["bin"] = pd.qcut(
        temp[feature],
        q=bins,
        duplicates="drop"
    )

    # Group data by bins
    grouped = temp.groupby(
        "bin",
        observed=False
    )[target].agg(
        ["count", "sum"]
    )

    # sum = number of lapse cases
    grouped["event"] = grouped["sum"]

    # non_event = number of non-lapse cases
    grouped["non_event"] = (
        grouped["count"] - grouped["event"]
    )

    # Total events and non-events
    total_event = grouped["event"].sum()
    total_non_event = grouped["non_event"].sum()

    # Distribution of events
    grouped["dist_event"] = (
        grouped["event"] / total_event
    )

    # Distribution of non-events
    grouped["dist_non_event"] = (
        grouped["non_event"] / total_non_event
    )

    # WOE calculation
    grouped["woe"] = np.log(
        (grouped["dist_non_event"] + 0.0001)
        /
        (grouped["dist_event"] + 0.0001)
    )

    # IV calculation
    grouped["iv"] = (
        grouped["dist_non_event"]
        -
        grouped["dist_event"]
    ) * grouped["woe"]

    # Total IV
    total_iv = grouped["iv"].sum()

    return total_iv
```

##### 5.2.2 Calculate IV for all numerical features


```python
iv_results = {}

for feature in numeric_df.columns:

    # Do not calculate IV for target
    if feature == "lapse_next_period":
        continue

    try:

        iv_value = calculate_iv(
            df,
            feature,
            "lapse_next_period"
        )

        iv_results[feature] = iv_value

    except Exception as e:

        print(
            "Could not calculate IV for:",
            feature
        )

# Convert IV result into DataFrame
iv_df = pd.DataFrame(
    list(iv_results.items()),
    columns=["Feature", "IV"]
)

# Sort from highest to lowest IV
iv_df = iv_df.sort_values(
    by="IV",
    ascending=False
)

print("\n========== INFORMATION VALUE ==========")
print(iv_df)
```

##### 5.3 WEIGHT OF EVIDENCE (WOE)


##### 5.3.1 Function to Calculate WOE Table


```python
def calculate_woe_table(data, feature, target, bins=10):

    temp = data[[feature, target]].copy()

    # Create bins
    temp["bin"] = pd.qcut(
        temp[feature],
        q=bins,
        duplicates="drop"
    )

    # Group by bins
    grouped = temp.groupby(
        "bin",
        observed=False
    )[target].agg(
        ["count", "sum"]
    )

    # Lapse customers
    grouped["event"] = grouped["sum"]

    # Non-lapse customers
    grouped["non_event"] = (
        grouped["count"] - grouped["event"]
    )

    # Total values
    total_event = grouped["event"].sum()
    total_non_event = grouped["non_event"].sum()

    # Event distribution
    grouped["event_dist"] = (
        grouped["event"] / total_event
    )

    # Non-event distribution
    grouped["non_event_dist"] = (
        grouped["non_event"] / total_non_event
    )

    # WOE
    grouped["WOE"] = np.log(
        (grouped["non_event_dist"] + 0.0001)
        /
        (grouped["event_dist"] + 0.0001)
    )

    return grouped
```

##### 5.8 SHAP FEATURE IMPORTANCE


```python
try:

    import shap
    import numpy as np
    import pandas as pd

    # Create SHAP Tree Explainer
    explainer = shap.TreeExplainer(rf_model)

    # Calculate SHAP values
    shap_values = explainer.shap_values(X_rf)

    print("\n========== SHAP CALCULATION COMPLETED ==========")

    # Check SHAP output
    print("SHAP output type:", type(shap_values))

    # ------------------------------------------------
    # Handle different SHAP versions
    # ------------------------------------------------

    if isinstance(shap_values, list):

        # Older SHAP versions
        # Class 0 = No Lapse
        # Class 1 = Lapse
        shap_class_1 = shap_values[1]

    else:

        shap_values_array = np.asarray(shap_values)

        print("SHAP array shape:", shap_values_array.shape)

        if shap_values_array.ndim == 3:

            # Newer SHAP versions
            # Shape = (samples, features, classes)

            shap_class_1 = shap_values_array[:, :, 1]

        elif shap_values_array.ndim == 2:

            # Shape = (samples, features)

            shap_class_1 = shap_values_array

        else:

            raise ValueError(
                f"Unexpected SHAP shape: {shap_values_array.shape}"
            )

    # ------------------------------------------------
    # Mean absolute SHAP importance
    # ------------------------------------------------

    shap_importance_values = np.abs(
        shap_class_1
    ).mean(axis=0)

    print(
        "SHAP importance shape:",
        shap_importance_values.shape
    )

    print(
        "Number of features:",
        len(X_rf.columns)
    )

    # ------------------------------------------------
    # Create SHAP DataFrame
    # ------------------------------------------------

    shap_df = pd.DataFrame({
        "Feature": X_rf.columns,
        "SHAP_Importance": shap_importance_values
    })

    # Sort by importance
    shap_df = shap_df.sort_values(
        by="SHAP_Importance",
        ascending=False
    ).reset_index(drop=True)

    # ------------------------------------------------
    # Display results
    # ------------------------------------------------

    print("\n========== SHAP FEATURE IMPORTANCE ==========")
    print(shap_df)

    print("\n========== TOP 15 SHAP FEATURES ==========")
    print(shap_df.head(15))


except ImportError:

    print(
        "\nSHAP is not installed."
        "\nInstall it using: %pip install shap"
    )
```

##### Step 10: Model Evaluation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    precision_recall_curve,
    auc,
    roc_curve
)

from scipy.stats import ks_2samp


models = {
    "Logistic Regression": best_log_model,
    "Decision Tree": best_dt_model,
    "Random Forest": best_rf_model,
    "AdaBoost": best_ada_model,
    "KNN": best_knn_model
}


evaluation_results = []

confusion_matrices = {}

pr_curves = {}

roc_curves = {}


for name, model in models.items():

    y_pred = model.predict(X_test)

    y_prob = model.predict_proba(X_test)[:, 1]

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )

    gini = (2 * roc_auc) - 1

    cm = confusion_matrix(
        y_test,
        y_pred
    )

    confusion_matrices[name] = cm

    precision_curve, recall_curve, _ = precision_recall_curve(
        y_test,
        y_prob
    )

    pr_auc = auc(
        recall_curve,
        precision_curve
    )

    pr_curves[name] = (
        recall_curve,
        precision_curve
    )

    fpr, tpr, _ = roc_curve(
        y_test,
        y_prob
    )

    roc_curves[name] = (
        fpr,
        tpr
    )

    lapse_scores = y_prob[y_test.values == 1]

    non_lapse_scores = y_prob[y_test.values == 0]

    ks_statistic = ks_2samp(
        lapse_scores,
        non_lapse_scores
    ).statistic

    evaluation_results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1,
        "ROC_AUC": roc_auc,
        "PR_AUC": pr_auc,
        "KS_Statistic": ks_statistic,
        "Gini": gini
    })


evaluation_df = pd.DataFrame(
    evaluation_results
)

evaluation_df = evaluation_df.sort_values(
    by="ROC_AUC",
    ascending=False
)

print("MODEL EVALUATION RESULTS")

print(
    evaluation_df.round(4)
)


for name, cm in confusion_matrices.items():

    plt.figure(figsize=(5, 4))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["No Lapse", "Lapse"],
        yticklabels=["No Lapse", "Lapse"]
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title(f"Confusion Matrix - {name}")
    plt.show()


plt.figure(figsize=(8, 6))

for name, values in roc_curves.items():

    fpr, tpr = values

    roc_auc = evaluation_df.loc[
        evaluation_df["Model"] == name,
        "ROC_AUC"
    ].values[0]

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC = {roc_auc:.3f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()


plt.figure(figsize=(8, 6))

for name, values in pr_curves.items():

    recall_curve, precision_curve = values

    pr_auc = evaluation_df.loc[
        evaluation_df["Model"] == name,
        "PR_AUC"
    ].values[0]

    plt.plot(
        recall_curve,
        precision_curve,
        label=f"{name} (PR-AUC = {pr_auc:.3f})"
    )

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.show()


print("STEP 10 COMPLETED SUCCESSFULLY")
```

##### Step 11 = Cross Validation.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import KFold, StratifiedKFold, cross_val_score

models = {
    "Logistic Regression": best_log_model,
    "Decision Tree": best_dt_model,
    "Random Forest": best_rf_model,
    "AdaBoost": best_ada_model,
    "KNN": best_knn_model
}

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

kfold_results = []

for name, model in models.items():

    scores = cross_val_score(
        model,
        X_train_balanced,
        y_train_balanced,
        cv=kf,
        scoring="roc_auc",
        n_jobs=-1
    )

    kfold_results.append({
        "Model": name,
        "Fold_1": scores[0],
        "Fold_2": scores[1],
        "Fold_3": scores[2],
        "Fold_4": scores[3],
        "Fold_5": scores[4],
        "Mean_ROC_AUC": scores.mean(),
        "Std_ROC_AUC": scores.std()
    })

kfold_df = pd.DataFrame(
    kfold_results
)

print("K-FOLD CROSS VALIDATION RESULTS")

print(
    kfold_df.round(4)
)

stratified_results = []

for name, model in models.items():

    scores = cross_val_score(
        model,
        X_train_balanced,
        y_train_balanced,
        cv=skf,
        scoring="roc_auc",
        n_jobs=-1
    )

    stratified_results.append({
        "Model": name,
        "Fold_1": scores[0],
        "Fold_2": scores[1],
        "Fold_3": scores[2],
        "Fold_4": scores[3],
        "Fold_5": scores[4],
        "Mean_ROC_AUC": scores.mean(),
        "Std_ROC_AUC": scores.std()
    })

stratified_df = pd.DataFrame(
    stratified_results
)

print("\nSTRATIFIED K-FOLD CROSS VALIDATION RESULTS")

print(
    stratified_df.round(4)
)

stratified_df = stratified_df.sort_values(
    by="Mean_ROC_AUC",
    ascending=False
)

print("\nSORTED STRATIFIED K-FOLD RESULTS")

print(
    stratified_df.round(4)
)

print("\nSTEP 11 COMPLETED SUCCESSFULLY")
```
